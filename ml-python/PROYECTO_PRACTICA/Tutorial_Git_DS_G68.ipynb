## 🛠️ Nivel 1: La Terminal en VS Code
La terminal es donde sucede la magia. Prueba estos comandos en la terminal de VS Code (Ctrl + ñ):

1.  **pwd**: ¿Dónde estoy parado?
2.  **ls**: ¿Qué archivos hay en esta carpeta?
3.  **cd data**: Métete a la carpeta de datos.
4.  **cd ..**: Sal de la carpeta.
5.  **pip list**: Mira qué librerías tienes instaladas.

## 🌳 Nivel 2: Git Local (Tu Máquina del Tiempo)
Git guarda versiones de tu código para que nunca pierdas nada.

### Paso 1: Inicializar el repositorio
Abre la terminal dentro de esta carpeta (`PROYECTO_PRACTICA`) y escribe:
```bash
git init
```

### Paso 2: El 'Área de Preparación'
Para decirle a Git que quieres guardar un archivo:
```bash
git add Tutorial_Git_DS_G68.ipynb
```
O para agregar todo:
```bash
git add .
```

### Paso 3: Crear un Punto de Guardado (Commit)
```bash
git commit -m "Mi primera versión del tutorial"
```
